## Step 4: Understand the Training Loop

Before implementing, let's understand what happens in one training step:

1. **Forward Pass**: Feed data through the model to get predictions
2. **Calculate Loss**: Measure how wrong the predictions are
3. **Zero Gradients**: Clear old gradient information
4. **Backward Pass**: Calculate gradients (how much to adjust weights)
5. **Optimizer Step**: Update model weights based on gradients

Here's a simple analogy: Imagine tuning a guitar. Forward pass = check if it sounds right. Loss = measure the wrongness. Backward pass = figure out which string to turn. Optimizer = actually turn the string.

## Step 5: Define Visualization Function

The visualization function is provided. This creates plots to visualize training progress.

```
def plot_metrics(train_losses, val_losses, train_accuracies, val_accuracies):
    """
    Creates two plots:
    1. Loss over time - ideally should decrease
    2. Accuracy over time - ideally should increase
    """
    epochs = range(1, len(train_losses) + 1)

    # Create a figure with two subplots
    plt.figure(figsize=(12, 5))

    # Plot 1: Loss
    plt.subplot(1, 2, 1)
    plt.plot(epochs, train_losses, label='Training Loss', marker='o')
    plt.plot(epochs, val_losses, label='Validation Loss', marker='s')

    plt.title('Training and Validation Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)

    # Plot 2: Accuracy
    plt.subplot(1, 2, 2)
    plt.plot(epochs, train_accuracies, label='Training Accuracy', marker='o')
    plt.plot(epochs, val_accuracies, label='Validation Accuracy', marker='s')

    plt.title('Training and Validation Accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy (%)')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.show()
```

## Step 6: Load and Prepare Data

Let's prepare the wine dataset step by step.

```
# STEP 1: Load the wine dataset
wine = load_wine()
X = wine.data  # Features (13 chemical properties)
y = wine.target  # Labels (0, 1, or 2 for wine types)

print(f"Dataset shape: X={X.shape}, y={y.shape}")
print(f"Number of wine samples: {X.shape[0]}")
print(f"Number of features: {X.shape[1]}")
print(f"Number of classes: {len(np.unique(y))}")
```

## Step 7: Normalize Features

Normalization helps the model train faster and more accurately.

```
# STEP 2: Normalize features (make them have mean=0 and std=1)
# This is important because neural networks work better with scaled data
scaler = StandardScaler()
X_normalized = scaler.fit_transform(X)

print("Before normalization - mean:", X.mean(axis=0)[:3], "... std:", X.std(axis=0)[:3])
print("After normalization - mean:", X_normalized.mean(axis=0)[:3], "... std:", X_normalized.std(axis=0)[:3])
```

## Step 8: Split Data into Training and Validation Sets

We use 80% of data for training and 20% for validation.

```
# STEP 3: Split into training (80%) and validation (20%) sets
# stratify=y ensures both sets have similar class distribution
X_train, X_val, y_train, y_val = train_test_split(
    X_normalized, y,
    test_size=0.2,        # 20% for validation
    random_state=42,      # For reproducibility
    stratify=y            # Keep class distribution balanced
)

print(f"Training set: X_train={X_train.shape}, y_train={y_train.shape}")
print(f"Validation set: X_val={X_val.shape}, y_val={y_val.shape}")
```

## Step 9: Convert to PyTorch Tensors

PyTorch requires data in tensor format.

```
# STEP 4: Convert numpy arrays to PyTorch tensors
# FloatTensor for features (continuous values), LongTensor for labels (integers)
X_train_tensor = torch.FloatTensor(X_train)
y_train_tensor = torch.LongTensor(y_train)
X_val_tensor = torch.FloatTensor(X_val)
y_val_tensor = torch.LongTensor(y_val)

print(f"X_train_tensor type: {X_train_tensor.dtype}, shape: {X_train_tensor.shape}")
print(f"y_train_tensor type: {y_train_tensor.dtype}, shape: {y_train_tensor.shape}")
```

## Step 10: Create Data Loaders

Data loaders batch the data and shuffle it for training.

```
# STEP 5: Create DataLoaders
# DataLoader batches the data and shuffles it during training
batch_size = 16

# Create datasets that pair features with labels
train_dataset = torch.utils.data.TensorDataset(X_train_tensor, y_train_tensor)
val_dataset = torch.utils.data.TensorDataset(X_val_tensor, y_val_tensor)

# Create data loaders
train_loader = torch.utils.data.DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True   # Shuffle training data for better learning
)
val_loader = torch.utils.data.DataLoader(
    val_dataset,
    batch_size=batch_size,
    shuffle=False  # Don't shuffle validation data
)

print(f"Number of batches in training set: {len(train_loader)}")
print(f"Number of batches in validation set: {len(val_loader)}")
```

## Step 11: Main Execution - Initialize and Train

Now let's set up the model, loss function, and optimizer, then train!

```
def main():
    # TODO: Set the hyperparameters
    # Your code here: Define learning_rate and num_epochs
    learning_rate = 0.01    # Controls how fast the model learns
    num_epochs = 50         # How many times to go through the training data

    # Initialize the model
    model = SimpleNet()

    # Loss function: Measures how wrong predictions are
    # CrossEntropyLoss is standard for multi-class classification
    criterion = nn.CrossEntropyLoss()

    # Optimizer: Updates model weights to minimize loss
    # Adam is a popular choice that works well in most cases
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)

    print("Model created successfully!")
    print(f"Learning rate: {learning_rate}")
    print(f"Number of epochs: {num_epochs}")
    print("-" * 60)

    # Train the model
    train_losses, val_losses, train_accuracies, val_accuracies = train_model(
        model, train_loader, val_loader, criterion, optimizer, num_epochs
    )

    # Plot the results
    plot_metrics(train_losses, val_losses, train_accuracies, val_accuracies)

    print("Training complete!")
```

## Step 12: Execute Training

Run the main function to train the model.

```
if __name__ == '__main__':
    main()
```
